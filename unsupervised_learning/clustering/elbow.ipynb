```python
import numpy as np
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist
import matplotlib.pyplot as plt

# Create two clusters
cluster1 = np.random.uniform(0.5, 1.5, (2, 10))
cluster2 = np.random.uniform(3.5, 4.5, (2, 10))

# Combine data
X = np.hstack((cluster1, cluster2)).T

# Range of K values
K = range(1, 10)

meandistortions = []

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    
    # compute distortion (distance to cluster centers)
    distortion = sum(
        np.min(cdist(X, kmeans.cluster_centers_, 'euclidean'), axis=1)
    )
    
    meandistortions.append(distortion)

# Plot Elbow Method
plt.plot(K, meandistortions, 'bx-')
plt.xlabel('Number of clusters (K)')
plt.ylabel('Distortion')
plt.title('Elbow Method for Optimal K')
plt.show()
```

```python
import numpy as np
from sklearn.cluster import KMeans
from scipy.spatial.distance import cdist
import matplotlib.pyplot as plt

# Create sample data (3 clusters idea)
cluster1 = np.random.uniform(0.5, 1.5, (2, 50))
cluster2 = np.random.uniform(3.5, 4.5, (2, 50))
cluster3 = np.random.uniform(6.0, 7.0, (2, 50))

X = np.hstack((cluster1, cluster2, cluster3)).T

# Try different K values
K = range(1, 10)
distortions = []

for k in K:
    kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)
    kmeans.fit(X)
    
    distortion = sum(
        np.min(cdist(X, kmeans.cluster_centers_, 'euclidean'), axis=1)
    )
    
    distortions.append(distortion)

# Plot elbow graph
plt.plot(K, distortions, 'bx-')
plt.xlabel("Number of clusters (K)")
plt.ylabel("Distortion")
plt.title("Elbow Method Graph")
plt.show()
```
