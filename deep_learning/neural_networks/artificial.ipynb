```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPClassifier
from sklearn.model_selection import KFold
import numpy as np  
```

```python
if __name__ == '__main__':
    # 1. Load the dataset
    digits = load_digits()
    X = digits.data
    y = digits.target
```

```python
kf = KFold(n_splits=3, shuffle=True, random_state=42)
scores = []

# Loop through each fold manually
for train_index, test_index in kf.split(X):
    # Split the data into train and test sets for this fold
    X_train, X_test = X[train_index], X[test_index]
    y_train, y_test = y[train_index], y[test_index]
    
    # 1. Fit the scaler ONLY on training data, then transform both
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)
    
    # 2. Initialize and fit the MLP Classifier
    mlp = MLPClassifier(hidden_layer_sizes=[150, 100], alpha=0.1, max_iter=500, random_state=42)
    mlp.fit(X_train_scaled, y_train)
    
    # 3. Evaluate the model on the scaled test data
    score = mlp.score(X_test_scaled, y_test)
    scores.append(score)

# Convert scores to a numpy array to match cross_val_score's output format
scores = np.array(scores)

print(f"Accuracies: {scores}")
```

```python
# Corrected pipeline definition
pipeline = Pipeline([
    ('ss', StandardScaler()),
    ('mlp', MLPClassifier(hidden_layer_sizes=[150, 100], alpha=0.1, max_iter=500, random_state=42))
])

# Running and printing the cross-validation scores
scores = cross_val_score(pipeline, X, y, n_jobs=-1, cv=3)
print(f"Accuracies: {scores}")
```
