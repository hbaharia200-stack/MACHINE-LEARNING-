```python
from sklearn.linear_model import LinearRegression

# Training data (Pizza diameter in inches)
X = [[6], [8], [10], [14], [18]]
# Target values (Price in dollars)
y = [[7], [9], [13], [17.5], [18]]

# Create and fit the model
model = LinearRegression()
model.fit(X, y)

# Prediction for a 12" pizza
# In modern sklearn, predict() expects a 2D array, so we use [[12]]
prediction = model.predict([[12]])

# Using a f-string (the modern Python way) for printing
print(f'A 12" pizza should cost: ${prediction[0][0]:.2f}')
```

We can visualize our training data by plotting it on a graph using matplotlib:The preceding script produces the following graph. The diameters of the pizzas areplotted on the x axis and the prices are plotted on the y axis.

```python
import matplotlib.pyplot as plt
X = [[6], [8], [10], [14], [18]]
y = [[7], [9], [13], [17.5], [18]]
plt.figure()
plt.title('Pizza price plotted against diameter')
plt.xlabel('Diameter in inches')
plt.ylabel('Price in dollars')
plt.plot(X, y, 'k.')
plt.axis([0, 25, 0, 25])
plt.grid(True)
plt.show()
```

```python
# No need for __future__ import in Python 3
x_values = [6, 8, 10, 14, 18]

# Calculate the mean (x-bar)
xbar = sum(x_values) / len(x_values)

# Calculate Sample Variance (divided by n-1, which is 4)
variance = sum((x - xbar)**2 for x in x_values) / (len(x_values) - 1)

print(f"The variance is: {variance}")
```

```python
import numpy as np

# Averages (Means)
xbar = (6 + 8 + 10 + 14 + 18) / 5
ybar = (7 + 9 + 13 + 17.5 + 18) / 5

# Manual Covariance Calculation
# Note: In Python 3, we don't need backslashes if the expression is inside parentheses
cov = ((6 - xbar) * (7 - ybar) + 
       (8 - xbar) * (9 - ybar) + 
       (10 - xbar) * (13 - ybar) +
       (14 - xbar) * (17.5 - ybar) + 
       (18 - xbar) * (18 - ybar)) / 4

print(f"Manual Covariance: {cov}")

# NumPy Covariance Calculation
# np.cov returns a matrix; [0][1] gives the covariance between the two sets
np_cov = np.cov([6, 8, 10, 14, 18], [7, 9, 13, 17.5, 18])[0][1]
print(f"NumPy Covariance: {np_cov}")
```

```python
# Kupata R-squared
r_sq = model.score(X, y)

print(f"R-squared: {r_sq:.4f}")
```
